```python
# ============================================================
# CELL 1: IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside Colab
%matplotlib inline

# Set a clean plotting style
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# ============================================================
# CELL 2: LOAD DATASET
# ============================================================

file_path = "SampleSuperstore_Excel_Dataset.xlsx"

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# ============================================================
# CELL 3: BASIC DATASET INFORMATION
# ============================================================

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types and Information:")
df.info()
```

```python
# ============================================================
# CELL 4: VIEW DATA
# ============================================================

print("First 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())
```

```python
# ============================================================
# CELL 5: CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# ============================================================
# CELL 6: CHECK DUPLICATES
# ============================================================

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

```python
# ============================================================
# CELL 7: SUMMARY STATISTICS
# ============================================================

print("Summary Statistics:")

display(df.describe())
```

```python
# ============================================================
# CELL 8: CATEGORICAL VARIABLES
# ============================================================

print("Category Distribution:")
display(df["Category"].value_counts())

print("\nRegion Distribution:")
display(df["Region"].value_counts())

print("\nSegment Distribution:")
display(df["Segment"].value_counts())

print("\nShip Mode Distribution:")
display(df["Ship Mode"].value_counts())
```

```python
# ============================================================
# CELL 9: CORRELATION HEATMAP
# ============================================================

numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap shows the relationships between Sales,
Quantity, Discount, and Profit. It helps identify positive and
negative relationships among the numerical variables.

```python
# ============================================================
# CELL 10: SALES BY CATEGORY
# ============================================================

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("\nSales by Category:")
display(category_sales)
```

### Observation

The chart compares the total sales generated by each product
category. The category with the highest bar contributes the
highest total sales.

```python
# ============================================================
# CELL 11: PROFIT BY CATEGORY
# ============================================================

category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("\nProfit by Category:")
display(category_profit)
```

### Observation

The chart compares the total profit generated by each product
category. The category with the highest bar contributes the
highest total profit.

```python
# ============================================================
# CELL 12: SALES BY REGION
# ============================================================

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("\nSales by Region:")
display(region_sales)
```

### Observation

Sales are distributed differently across the four regions.
The region with the highest bar contributes the largest total
sales.

```python
# ============================================================
# CELL 13: SALES DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

plt.hist(
    df["Sales"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### Observation

The histogram shows the distribution of sales values. Most
transactions are concentrated in the lower sales range, while
a smaller number of transactions have much higher sales values.

```python
# ============================================================
# CELL 14: PROFIT DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

plt.hist(
    df["Profit"],
    bins=30,
    edgecolor="black"
)

plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### Observation

The histogram shows the distribution of profit values. It helps
identify the common profit range as well as unusually high or
low profit values.

```python
# ============================================================
# CELL 15: SALES VS PROFIT
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    alpha=0.6
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

### Observation

The scatter plot shows the relationship between Sales and Profit.
It also helps identify transactions where sales are high but
profit is relatively low or negative.

```python
# ============================================================
# CELL 16: PROFIT OUTLIER ANALYSIS
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Profit"]
)

plt.title("Profit Distribution and Potential Outliers")
plt.xlabel("Profit")

plt.tight_layout()
plt.show()
```

### Observation

The boxplot shows the central distribution of profit
and highlights potential outliers. These extreme values may
represent unusually profitable or loss-making transactions.

```python
# ============================================================
# CELL 16A: STATISTICAL OUTLIER DETECTION USING IQR
# ============================================================

Q1 = df["Profit"].quantile(0.25)
Q3 = df["Profit"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Profit"] < lower_bound) |
    (df["Profit"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Profit Outliers:", len(outliers))
```

```python
# ============================================================
# CELL 17: CATEGORY AND REGION SUMMARY
# ============================================================

summary = df.groupby(
    ["Region", "Category"]
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Sales=("Sales", "mean"),
    Average_Profit=("Profit", "mean")
).reset_index()

display(summary)
```

```python
# ============================================================
# CELL 18: TOP 10 CITIES BY SALES
# ============================================================

top_cities = (
    df.groupby("City")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))

top_cities.sort_values().plot(kind="barh")

plt.title("Top 10 Cities by Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("City")

plt.tight_layout()
plt.show()

print("Top 10 Cities by Sales:")
display(top_cities)
```

```python
# ============================================================
# CELL 19: TOP 3 INSIGHTS
# ============================================================

top_sales_category = category_sales.idxmax()
top_sales_category_value = category_sales.max()

top_profit_category = category_profit.idxmax()
top_profit_category_value = category_profit.max()

top_sales_region = region_sales.idxmax()
top_sales_region_value = region_sales.max()

print("=" * 60)
print("                    TOP 3 INSIGHTS")
print("=" * 60)

print("\n1. CATEGORY PERFORMANCE")
print(
    f"{top_sales_category} has the highest total sales, "
    f"with approximately {top_sales_category_value:,.2f} in sales."
)

print("\n2. PROFIT PERFORMANCE")
print(
    f"{top_profit_category} generates the highest total profit, "
    f"with approximately {top_profit_category_value:,.2f} in profit."
)

print("\n3. REGIONAL PERFORMANCE")
print(
    f"{top_sales_region} has the highest total sales among the "
    f"four regions, with approximately {top_sales_region_value:,.2f}."
)
```
